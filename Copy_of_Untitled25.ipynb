```python
from google.colab import files
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# Upload Concrete_Data.xls
uploaded = files.upload()

# Read the uploaded Excel file
file_name = next(iter(uploaded))
df = pd.read_excel(file_name)

# Clean column names for easy use
df.columns = [
    "Cement",
    "Blast_Furnace_Slag",
    "Fly_Ash",
    "Water",
    "Superplasticizer",
    "Coarse_Aggregate",
    "Fine_Aggregate",
    "Age",
    "Compressive_Strength"
]

# Remove duplicate records
df = df.drop_duplicates().copy()

# Display dataset details
print("Dataset Shape:", df.shape)
print("\nDataset Information:")
display(df.head())

print("\nSummary Statistics:")
display(df.describe().round(2))
```

```python
# Create age groups
df["Age_Group"] = pd.cut(
    df["Age"],
    bins=[0, 7, 28, 90, float("inf")],
    labels=["1-7 Days", "8-28 Days", "29-90 Days", "Above 90 Days"],
    include_lowest=True
)
```

```python
# Calculate average strength
grouped = df.groupby(
    "Age_Group", observed=False
)["Compressive_Strength"].mean().reset_index()
```

```python
# Create bar graph
plt.figure(figsize=(9, 5))
sns.barplot(data=grouped, x="Age_Group", y="Compressive_Strength")

plt.title("Average Concrete Compressive Strength by Age Group")
plt.xlabel("Curing Age Group")
plt.ylabel("Average Compressive Strength (MPa)")
plt.xticks(rotation=15)
plt.tight_layout()
plt.show()
```

```python
# Display grouped statistics
print("CONCRETE STRENGTH BY AGE GROUP")
display(
    df.groupby("Age_Group", observed=False)["Compressive_Strength"]
    .agg(["count", "mean", "median", "min", "max", "std"])
    .round(2)
)
```

```python
plt.figure(figsize=(9, 5))

sns.histplot(
    data=df,
    x="Compressive_Strength",
    bins=25,
    kde=True
)

plt.title("Distribution of Concrete Compressive Strength")
plt.xlabel("Compressive Strength (MPa)")
plt.ylabel("Frequency")
plt.tight_layout()
plt.show()
```

```python
# Select numerical columns only
numeric_df = df.drop(columns=["Age_Group"], errors="ignore")

# Calculate correlation
correlation = numeric_df.corr()
```

```python
# Display correlation values
display(correlation.round(2))
```

```python
# Draw heatmap
plt.figure(figsize=(12, 8))

sns.heatmap(
    correlation,
    annot=True,
    cmap="coolwarm",
    fmt=".2f",
    linewidths=0.5
)
```

```python
plt.figure(figsize=(9, 5))

sns.scatterplot(
    data=df,
    x="Cement",
    y="Compressive_Strength"
)

plt.title("Relationship Between Cement and Concrete Strength")
plt.xlabel("Cement Content (kg/m³)")
plt.ylabel("Compressive Strength (MPa)")
plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(9, 5))

sns.scatterplot(
    data=df,
    x="Age",
    y="Compressive_Strength",
    alpha=0.6
)

plt.title("Curing Age versus Concrete Compressive Strength")
plt.xlabel("Age (Days)")
plt.ylabel("Compressive Strength (MPa)")
plt.tight_layout()
plt.show()
```

```python
# Calculate average strength at each recorded age
age_strength = (
    df.groupby("Age")["Compressive_Strength"]
    .mean()
    .reset_index()
    .sort_values("Age")
)

plt.figure(figsize=(11, 5))

sns.lineplot(
    data=age_strength,
    x="Age",
    y="Compressive_Strength",
    marker="o"
)

plt.title("Average Concrete Strength Across Curing Ages")
plt.xlabel("Curing Age (Days)")
plt.ylabel("Average Compressive Strength (MPa)")
plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(9, 5))

sns.boxplot(
    data=df,
    x="Age_Group",
    y="Compressive_Strength"
)

plt.title("Concrete Strength Distribution by Age Group")
plt.xlabel("Curing Age Group")
plt.ylabel("Compressive Strength (MPa)")
plt.xticks(rotation=15)
plt.tight_layout()
plt.show()
```

```python
ingredients = [
    "Cement",
    "Blast_Furnace_Slag",
    "Fly_Ash",
    "Water",
    "Superplasticizer",
    "Coarse_Aggregate",
    "Fine_Aggregate"
]

mean_ingredients = df[ingredients].mean().sort_values(ascending=False)

plt.figure(figsize=(11, 6))

sns.barplot(
    x=mean_ingredients.values,
    y=mean_ingredients.index
)

plt.title("Average Quantity of Concrete Ingredients")
plt.xlabel("Average Quantity (kg/m³)")
plt.ylabel("Concrete Ingredient")
plt.tight_layout()
plt.show()
```

```python
print("FINAL DATA ANALYSIS SUMMARY")
print("-" * 40)

print("Total records after removing duplicates:", len(df))
print("Total columns including Age_Group:", len(df.columns))

print("\nAverage Compressive Strength:",
      round(df["Compressive_Strength"].mean(), 2), "MPa")

print("Median Compressive Strength:",
      round(df["Compressive_Strength"].median(), 2), "MPa")

print("Minimum Compressive Strength:",
      round(df["Compressive_Strength"].min(), 2), "MPa")

print("Maximum Compressive Strength:",
      round(df["Compressive_Strength"].max(), 2), "MPa")

print("\nCorrelation with Compressive Strength:")

strength_corr = (
    df.drop(columns=["Age_Group"], errors="ignore")
    .corr()["Compressive_Strength"]
    .drop("Compressive_Strength")
    .sort_values(ascending=False)
)

display(strength_corr.round(3))

print("\nStrongest positive correlation:",
      strength_corr.idxmax(),
      "=", round(strength_corr.max(), 3))

print("Strongest negative correlation:",
      strength_corr.idxmin(),
      "=", round(strength_corr.min(), 3))
```
